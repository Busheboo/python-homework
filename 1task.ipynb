```python
class Chocolate:
    def __init__(self, price: float, additive: str = None):
        # Цена должна быть больше нуля
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = price
        self.additive = additive

    def show_my_chocolate(self):
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"
```

```python
# Батончик с добавкой
choco1 = Chocolate(55.5, "орехи")
choco1.show_my_chocolate()
print(choco1)

# Обычный батончик (без добавки)
choco2 = Chocolate(40.0)
choco2.show_my_chocolate()
print(choco2)

# Еще один с добавкой
choco3 = Chocolate(70, "карамель")
choco3.show_my_chocolate()
print(choco3)

# Проверка исключения
try:
    choco4 = Chocolate(-10, "изюм")
except ValueError as e:
    print("Ошибка:", e)

try:
    choco5 = Chocolate(0)
except ValueError as e:
    print("Ошибка:", e)
```
