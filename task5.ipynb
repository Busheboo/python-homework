Часть 3

```python
housing = pd.read_csv('Housing.csv')
housing.head()
```

```python
housing.info(memory_usage='deep')
print(housing.dtypes)
before = housing.memory_usage(deep=True).sum()
print('Память до оптимизации, байт:', before)
```

```python
housing.describe(include='all').T
```

```python
# для категориальных столбцов смотрим уникальные значения
for col in housing.select_dtypes(include='object').columns:
    print(col, housing[col].unique())
```

```python
print(housing[['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']].agg(['min', 'max']))

# Выводы по оптимальным типам:
# price: значения от 1,75 млн до 13,3 млн, не помещаются в int16, но помещаются в int32 -> int32
# area: значения от 1650 до 16200, помещаются в int16 (до 32767) -> int16
# bedrooms (1..6), bathrooms (1..4), stories (1..4), parking (0..3):
#   малые целые числа, помещаются в int8 (до 127) -> int8
# mainroad, guestroom, basement, hotwaterheating, airconditioning, prefarea:
#   только два значения 'yes'/'no' -> bool
# furnishingstatus: 3 уникальных строковых значения -> category
```

```python
housing_opt = housing.copy()

assert housing['area'].max() < 32768   # проверка, что int16 не переполнится

housing_opt['price'] = housing_opt['price'].astype('int32')
housing_opt['area'] = housing_opt['area'].astype('int16')

for col in ['bedrooms', 'bathrooms', 'stories', 'parking']:
    housing_opt[col] = housing_opt[col].astype('int8')

bool_cols = ['mainroad', 'guestroom', 'basement',
             'hotwaterheating', 'airconditioning', 'prefarea']
for col in bool_cols:
    housing_opt[col] = housing_opt[col].map({'yes': True, 'no': False})

housing_opt['furnishingstatus'] = housing_opt['furnishingstatus'].astype('category')

after = housing_opt.memory_usage(deep=True).sum()
print(housing_opt.dtypes)
print('Память до:', before, 'байт')
print('Память после:', after, 'байт')
print(f'Экономия: {(1 - after / before) * 100:.1f}%')
```
